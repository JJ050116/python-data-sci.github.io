# 📊 Week 05 — Analysis พื้นฐาน (Properties, Methods, Functions)

การวิเคราะห์ข้อมูลคือการเปลี่ยนข้อมูลรายแถวให้กลายเป็นคำตอบภาพรวม ในบทเรียน Week 05 นี้ เราจะเรียนรู้ตัวอย่างวิธีการใช้งาน **Properties**, **Methods**, และ **Functions** สำหรับวิเคราะห์ข้อมูลด้วย Pandas


## 📌 สรุปประเภทของคำสั่งใน Python & Pandas

- **Properties (Attributes)** = ข้อมูลคุณลักษณะประจำตัวของ Object **ไม่ต้องใส่ `()`** 
  - *ตัวอย่าง:* `.shape`, `.columns`, `.index`
- **Methods** = ฟังก์ชันที่สังกัดอยู่กับ Object เรียกใช้โดยต่อท้ายจุด `.` และ **ต้องใส่ `()`**
  - *ตัวอย่าง:* `df.head()`, `df.query()`, `df.groupby()`, `.mean()`, `.count()`, `.agg()`, `.sort_values()`, `.reset_index()`, `.value_counts()`, `.describe()`
- **Functions** = ฟังก์ชันอิสระ เรียกใช้งานโดยส่งข้อมูลผ่าน **`()`** 
  - *ตัวอย่าง Built-in / Library:* `pd.read_csv()`, `display()`, `print()`, `len()`, `round()`
  - *ตัวอย่าง Custom Function:* การใช้คำสั่ง `def` เพื่อเขียนฟังก์ชันวิเคราะห์ข้อมูลขึ้นมาใช้เอง


```python
import pandas as pd

# โหลดชุดข้อมูลนักเตะ footballers.csv
df = pd.read_csv("../data/footballers.csv")
display(df.head())
```

---
## 1️⃣ ตัวอย่างการใช้งาน Properties (ไม่มี `()`)
Properties คือข้อมูลคุณลักษณะของ DataFrame หรือ Series ที่อ่านค่าได้ทันทีโดยไม่ต้องใส่ `()`


```python
print("--- Properties Examples ---")
print("1. df.shape (จำนวนแถว, จำนวนคอลัมน์):", df.shape)
print("2. df.columns (รายชื่อคอลัมน์ทั้งหมด):", df.columns.tolist())
print("3. df.index (ประเภทของ Index):", df.index)
```

---
## 2️⃣ ตัวอย่างการใช้งาน Methods ที่สอนใน Week 05 (ต้องมี `()`)
Methods คือคำสั่งที่กระทำกับ DataFrame/Series โดยตรงเพื่อทำการกรอง สรุปผล หรือจัดกลุ่มข้อมูล


```python
# 2.1 .groupby() + .mean()
# แบ่งกลุ่มนักเตะตาม league แล้วหาค่า overall เฉลี่ย
league_overall_mean = df.groupby("league")["overall"].mean()

print("=== 2.1 ค่า overall เฉลี่ยแยกตาม league (.groupby() + .mean()) ===")
display(league_overall_mean)
```

```python
# 2.2 .sort_values() + .reset_index()
# เรียงลำดับจากมากไปน้อย (ascending=False) และแปลง Index กลับเป็นคอลัมน์ปกติ
sorted_league = league_overall_mean.sort_values(ascending=False).reset_index()
sorted_league.columns = ["league", "average_overall"]

print("=== 2.2 ผลลัพธ์หลังเรียงลำดับ (.sort_values) และแปลงเป็น DataFrame (.reset_index) ===")
display(sorted_league.round(2))
```

```python
# 2.3 .agg()
# สรุปสถิติหลายค่าพร้อมกัน (count, mean, median) ของ overall แยกตาม position
position_stats = df.groupby("position")["overall"].agg(["count", "mean", "median"]).sort_values(by="mean", ascending=False)

print("=== 2.3 สรุปคะแนน overall นักเตะแยกตามตำแหน่ง (.groupby() + .agg()) ===")
display(position_stats.round(1))
```

```python
# 2.4 .value_counts() & .describe()
print("=== 2.4.1 จำนวนนักเตะแยกตามสัญชาติ Top 5 (.value_counts()) ===")
display(df["nationality"].value_counts().head(5))

print()
print("=== 2.4.2 สถิติเชิงพรรณนาภาพรวมของคอลัมน์ตัวเลข (.describe()) ===")
display(df[["age", "overall", "pace", "shooting", "passing"]].describe().round(1))
```

---
## 3️⃣ ตัวอย่างการใช้งาน Functions (ส่งค่าเข้าใน `()`)
Functions เป็นคำสั่งที่รับ Input ผ่าน `()` เพื่อนำไปประมวลผล รวมถึงการเขียน Custom Function ด้วย `def`


```python
# 3.1 Built-in & Library Functions (เช่น len(), round(), display(), print())
total_players = len(df) # len() นับจำนวนแถวใน DataFrame
avg_age = round(df["age"].mean(), 1) # round() ปัดเศษทศนิยม

print(f"จำนวนนักเตะทั้งหมด (ใช้ len()): {total_players} คน")
print(f"อายุเฉลี่ยของนักเตะ (ใช้ round()): {avg_age} ปี")
```

```python
# 3.2 Custom Function (สร้างฟังก์ชันขึ้นมาใช้งานเองด้วย def)
def analyze_group_stats(data, group_col, target_col):
    """
    ฟังก์ชันแบ่งกลุ่มข้อมูล คำนวณหา count, mean, median
    และเรียงลำดับค่าเฉลี่ยจากมากไปน้อย
    """
    result = data.groupby(group_col)[target_col].agg(["count", "mean", "median"])
    result = result.sort_values(by="mean", ascending=False).round(1)
    return result

# เรียกใช้งาน Custom Function analyze_group_stats
print("=== ผลลัพธ์จากการเรียกใช้ Custom Function: analyze_group_stats() ===")
league_pace_summary = analyze_group_stats(df, group_col="league", target_col="pace")
display(league_pace_summary)
```
