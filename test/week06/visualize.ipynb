```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../data/footballers.csv")
display(df.head(3))
```

```python
top_5 = df.nlargest(5, "overall")[["name", "overall", "value_millions"]]
display(top_5)
```

```python
# player = df.query("name == 'Alphonso Davies' ")
# display(player)
plt.figure(figsize=(8, 4.5))
plt.bar(top_5["name"], top_5["overall"], color="green", edgecolor="black")

plt.title("Top 5 player")
plt.xlabel("Player name")
plt.ylabel("Overall")

plt.xticks(rotation=20)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.savefig("Top 5 player overall rating")
plt.show()

print("Save figure successfully....")
```
