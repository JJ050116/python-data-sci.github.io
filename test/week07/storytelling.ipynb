# Week07 - Data Story telling & Report

```python
import pandas as pd

df = pd.read_csv("../../data/footballers.csv")
display(df.head(3))
```

```python
league_report = (
    df.groupby("league")["overall"]
    .agg(player_count="count", avg_overall="mean")
    .round(2)
    .sort_values(by="avg_overall", ascending=False)
    .reset_index()
)

print("=== Table for Board ===")
display(league_report)
```

### section 2

```python
top_league = league_report.iloc[0]
sample_size = len(df)

sentence_1 = f"Info player league top score: {top_league['league']} league"
sentence_2 = f"Score overall avg: {top_league['avg_overall']} from player in this league"
sentence_3 = f"Only: {sample_size} player in 5 league"

full_story = f"""
summarize data: 
1. {sentence_1}
2. {sentence_2}
3. {sentence_3}
"""

print(full_story)
```
