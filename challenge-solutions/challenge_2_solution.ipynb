## ภารกิจ
สร้างตัวแปร `summary` ที่มีเพียง `name`, `club`, `overall` แล้วเปลี่ยนชื่อ `overall` เป็น `rating` โดยที่ `df` ต้นฉบับต้องไม่เปลี่ยนแปลง

### 🔍 วิเคราะห์ Debug Checkpoint
เมื่อรัน `df[["name", "club", "Overall"]]` จะเกิด `KeyError: 'Overall'`
- **สาเหตุ:** Python และ Pandas เป็น Case-Sensitive (ตัวพิมพ์เล็ก-ใหญ่มีความหมายต่างกัน) ในไฟล์จริงคอลัมน์ชื่อ `overall` ตัวพิมพ์เล็กทั้งหมด
- **วิธีแก้:** ตรวจสอบชื่อคอลัมน์จริงด้วย `df.columns` ก่อน แล้วระบุชื่อให้ถูกต้อง

```python
import pandas as pd

df = pd.read_csv("../data/footballers.csv")

# 1. เลือกเฉพาะ 3 คอลัมน์ที่ต้องการ
selected_columns = ["name", "club", "overall"]
summary = df[selected_columns].copy()

# 2. เปลี่ยนชื่อคอลัมน์ overall เป็น rating
summary = summary.rename(columns={"overall": "rating"})

# 3. แสดง 5 แถวแรกของ summary
display(summary.head(5))

# 4. ตรวจสอบว่า df ต้นฉบับยังคงเดิม
print("คอลัมน์ของ summary:", list(summary.columns))
print("คอลัมน์ของ df เดิม:", list(df.columns[:5]))
```

### 💡 คำอธิบายเพิ่มเติม
- การใช้ `df[[...]]` สองชั้น เพื่อให้ได้ผลลัพธ์เป็น **DataFrame (ตาราง)** ไม่ใช่ Series (คอลัมน์เดี่ยว)
- การใช้ `.copy()` ช่วยให้แน่ใจว่าการแก้ไขใน `summary` จะเป็นตารางอิสระ ไม่ส่งผลกระทบต่อ `df` เดิม
