## ภารกิจ
กรองหานักเตะที่มีคุณสมบัติตรงตาม 2 เงื่อนไขพร้อมกัน:
1. ตำแหน่ง (`position`) เป็น `ST` หรือ `RW`
2. ความแข็งแกร่ง (`physical`) ตั้งแต่ `75` ขึ้นไป
3. แสดงผลเฉพาะคอลัมน์ `name`, `position`, `physical`

### 🔍 วิเคราะห์ Debug Checkpoint
ทำไมการใช้คำว่า `and` ใน Python ถึงพัง (`ValueError: The truth value of a Series is ambiguous`)?
- คำว่า `and` ใน Python ใช้เปรียบเทียบค่าความจริงเดี่ยว (Single Boolean True/False) แต่ `df["position"].isin(...)` ส่งค่ากลับมาเป็น Series ที่มี True/False หลายแถว
- **วิธีแก้ใน Pandas:** ต้องใช้เครื่องหมาย **`&` (Bitwise AND)** เพื่อประเมินค่าทีละแถว (Element-wise) และต้องครอบวงเล็บ `(...) & (...)` รอบแต่ละเงื่อนไขเสมอ

```python
import pandas as pd

df = pd.read_csv("../data/footballers.csv")

# ==============================================================================
# วิธีที่ 1: ใช้ Boolean Masking ด้วย & (วิธีมาตรฐานใน Pandas)
# ==============================================================================
# 1. isin(["ST", "RW"]) คืนค่าเป็น Series ของ True/False
is_forward = df["position"].isin(["ST", "RW"])

# 2. physical >= 75 คืนค่าเป็น Series ของ True/False
is_strong = df["physical"] >= 75

# 3. นำ 2 เงื่อนไขมารวมกันด้วย & (Bitwise AND)
# - ใช้ & แทน 'and' เพราะต้องการประเมินความจริงทีละแถว (Element-wise)
# - ต้องมีวงเล็บ () ครอบเงื่อนไขย่อยเสมอ เพราะ & มีลำดับความสำคัญสูงกว่าเครื่องหมายเปรียบเทียบ
# - [["name", "position", "physical"]] เป็นการเลือกคอลัมน์เฉพาะที่ต้องการ
strong_forwards = df[is_forward & is_strong][["name", "position", "physical"]]
display(strong_forwards)
```

```python
# ==============================================================================
# วิธีที่ 2 (ทางเลือก): ใช้คำสั่ง query()
# ==============================================================================
# - ใช้ query() เมื่อต้องการเขียนเงื่อนไขด้วย String ภาษาพูด/SQL
# - ใน String ของ query() สามารถใช้คำว่า 'and', 'or', 'in' ได้โดยตรง (ไม่ต้องใช้ & และไม่ต้องใส่ () )
query_result = df.query("position in ['ST', 'RW'] and physical >= 75")[["name", "position", "physical"]]
display(query_result)
```

### 💡 คำอธิบายเพิ่มเติม
- มีนักเตะเพียงบางคนที่ผ่านเกณฑ์ เพราะเงื่อนไข `&` บังคับให้ต้องเป็นจริง **ทั้งสองข้อ** (เป็นกองหน้า และแข็งแกร่งด้วย)
