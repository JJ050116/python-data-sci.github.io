## ภารกิจ
1. ใช้ `groupby` เพื่อแบ่งกลุ่มข้อมูลตามสัญชาติ (`nationality`)
2. คำนวณหาค่าความเร็วเฉลี่ย (`pace` mean)
3. เรียงลำดับจากค่าเฉลี่ยมากไปน้อย
4. เขียนข้อสรุปที่คำนึงถึงขนาดกลุ่มตัวอย่าง

### 🔍 วิเคราะห์ Debug Checkpoint
หากพิมพ์ `df.groupby("nationalities")` จะเกิด `KeyError: 'nationalities'` เพราะชื่อคอลัมน์จริงในไฟล์เป็นรูปเอกพจน์ (`nationality`)
- **ข้อคิดทางสถิติ:** ค่าเฉลี่ยของประเทศที่มีนักเตะเพียง 1 คน (เช่น Morocco, Canada, France) อาจดูสูงมาก แต่ไม่ได้แปลว่านักเตะทุกคนในประเทศนั้นวิ่งเร็วเหมือนกัน จึงควรตรวจ `count` ร่วมด้วย

```python
import pandas as pd

df = pd.read_csv("../data/footballers.csv")

# 1. คำนวณ pace เฉลี่ยแยกลำดับสัญชาติ
nationality_pace = df.groupby("nationality")["pace"].mean().sort_values(ascending=False)

print("=== Pace เฉลี่ยแยกตามประเทศ (Top 10) ===")
display(nationality_pace.head(10).round(1))

# 2. ดู count ควบคู่กับ mean เพื่อเห็นภาพจริง
stats = df.groupby("nationality")["pace"].agg(["count", "mean"]).sort_values("mean", ascending=False)
print("\n=== สถิติแบบละเอียด (Count + Mean) ===")
display(stats.head(10).round(1))
```
