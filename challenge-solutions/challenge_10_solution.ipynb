## 1. คำถามวิจัย (Research Question)
> **"ตำแหน่งการเล่นใด (Position) ในหมู่นักเตะระดับท็อป มีมูลค่าประเมินเฉลี่ย (`value_millions`) สูงที่สุด และมีรูปแบบการกระจายตัวอย่างไร?"**

```python
# 2. เตรียม Library และทำความสะอาดข้อมูล
import pandas as pd
import matplotlib.pyplot as plt

# 2.1 โหลดข้อมูล
df = pd.read_csv("../data/footballers.csv")

# 2.2 จัดการชนิดข้อมูลและค่าที่หายไป (Data Cleaning)
df["value_millions"] = pd.to_numeric(df["value_millions"], errors="coerce")
clean_df = df.dropna(subset=["value_millions"]).copy()

print(f"จำนวนแถวก่อน clean: {len(df)} แถว")
print(f"จำนวนแถวหลัง clean: {len(clean_df)} แถว (ตัดค่า unknown ออก {len(df) - len(clean_df)} แถว)")
```

```python
# 3. วิเคราะห์ข้อมูล (Data Analysis)
# คำนวณค่าเฉลี่ย, ค่ามัธยฐาน และจำนวนนักเตะในแต่ละตำแหน่ง
position_stats = (
    clean_df.groupby("position")["value_millions"]
    .agg(count="count", mean="mean", median="median")
    .sort_values("mean", ascending=False)
    .round(1)
)

print("=== ตารางสรุปสถิติมูลค่านักเตะแยกตามตำแหน่ง ===")
display(position_stats)
```

```python
# 4. แสดงผลด้วยกราฟ (Data Visualization)
plt.figure(figsize=(9, 5))

bars = plt.bar(
    position_stats.index, 
    position_stats["mean"], 
    color="#2a9d8f", 
    edgecolor="#1d3557",
    linewidth=1.2
)

# เพิ่มตัวเลขบนแท่งกราฟเพื่อให้อ่านค่าง่าย
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2, 
        height + 2, 
        f"{height:.0f}M", 
        ha="center", 
        va="bottom", 
        fontsize=10,
        fontweight="bold"
    )

plt.title("Average Player Market Value by Position", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Position (ตำแหน่ง)", fontsize=11)
plt.ylabel("Average Value (Millions of Euros)", fontsize=11)
plt.ylim(0, position_stats["mean"].max() + 20)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
