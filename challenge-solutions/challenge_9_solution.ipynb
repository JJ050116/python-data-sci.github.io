วางแผนโครงการ Data Science ขนาดเล็ก (Mini Project) โดยครอบคลุม 4 ส่วนสำคัญก่อนลงมือเขียนโค้ดวิเคราะห์

### 📋 ตัวอย่างแผนงาน Mini Project (4 ส่วน)

1. **คำถาม (Question):**
   > *"ตำแหน่งการเล่นใด (Position) ใน 5 ลีกใหญ่ มีมูลค่านักเตะเฉลี่ย (`value_millions`) สูงที่สุด?"*

2. **คอลัมน์ที่ต้องใช้ (Columns):**
   > `position`, `value_millions`, `name`

3. **จุดที่ต้องตรวจและ Clean ข้อมูล (Data Cleaning):**
   > - ตรวจสอบว่า `value_millions` มีค่าข้อความ (เช่น `'unknown'`) หรือค่าว่าง (NaN) หรือไม่
   > - แปลงเป็นตัวเลขด้วย `pd.to_numeric` และตัดแถวว่างออก
   > - ตรวจจำนวนนักเตะในแต่ละตำแหน่งว่ามีกลุ่มตัวอย่างเพียงพอหรือไม่

4. **กราฟที่เลือกและเหตุผล (Visualization):**
   > - **Bar Chart (กราฟแท่ง):** เหมาะที่สุดสำหรับการเปรียบเทียบค่าเฉลี่ยตัวเลขระหว่างหมวดหมู่ตำแหน่งต่าง ๆ พร้อมแสดงจำนวนนักเตะ (Count) กำกับ

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. โหลดข้อมูล
df = pd.read_csv("../data/footballers.csv")

# 2. ตรวจสอบคุณภาพข้อมูลของคอลัมน์ที่เลือก
print("=== ตรวจสอบชนิดข้อมูลและ Missing Values ===")
df["value_millions"] = pd.to_numeric(df["value_millions"], errors="coerce")
clean_df = df.dropna(subset=["value_millions"]).copy()

# 3. สำรวจจำนวนต่อตำแหน่ง
print("\n=== จำนวนนักเตะแยกตามตำแหน่ง ===")
display(clean_df["position"].value_counts())

# 4. ทดลองคำนวณเบื้องต้น
position_value = clean_df.groupby("position")["value_millions"].agg(["count", "mean"]).sort_values("mean", ascending=False)
print("\n=== มูลค่าเฉลี่ยแยกตามตำแหน่ง ===")
display(position_value.round(1))
```
