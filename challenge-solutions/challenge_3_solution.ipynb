## ภารกิจ
ตรวจหาค่าที่หายไป (Missing value / NaN) ในคอลัมน์ `age` แล้วสร้าง `clean_df` ที่แทนค่าว่างด้วย **ค่ามัธยฐาน (Median)** โดยไม่แก้ไข `df` เดิม

### 🔍 วิเคราะห์ Debug Checkpoint
ทำไมจึงไม่ควรแทนค่า `age` ที่หายไปด้วย `0`?
- อายุ `0` ปีไม่มีจริงในหมู่นักฟุตบอลอาชีพ
- การใส่ `0` จะดึงค่าเฉลี่ยของอายุให้ต่ำลงอย่างผิดปกติ ทำให้ผลการวิเคราะห์ในสัปดาห์ถัดไปเพี้ยนทันที
- **Median (ค่ามัธยฐาน)** เป็นค่ากึ่งกลางของข้อมูลจริง จึงเป็นตัวแทนที่สมเหตุสมผลและทนต่อค่าผิดปกติ (Outliers) ได้ดี

```python
import pandas as pd

df = pd.read_csv("../data/footballers.csv")

# 1. ตรวจสอบจำนวนค่าที่หายไปในคอลัมน์ age
missing_before = df["age"].isna().sum()
print(f"จำนวนค่า age ที่หายไปก่อน clean: {missing_before} จุด")

# 2. คำนวณค่ามัธยฐาน (Median) ของ age
median_age = df["age"].median()
print(f"ค่ามัธยฐานของ age: {median_age:.1f} ปี")

# 3. สร้าง clean_df โดยแทนค่าหายด้วย median
clean_df = df.assign(age=df["age"].fillna(median_age))

# 4. ตรวจสอบยืนยันผล
missing_after = clean_df["age"].isna().sum()
print(f"จำนวนค่า age ที่หายไปใน clean_df: {missing_after} จุด")
print(f"จำนวนค่า age ที่หายไปใน df เดิม: {df['age'].isna().sum()} จุด (df เดิมไม่เปลี่ยน)")
```

### 💡 คำอธิบายเพิ่มเติม
- `df.assign(...)` ช่วยสร้าง DataFrame ใหม่ที่มีการปรับค่าคอลัมน์ โดยไม่กระทบกับตารางเดิมในหน่วยความจำ
- การตรวจผลด้วย `.isna().sum()` ทั้งก่อนและหลังแก้ เป็นหัวใจสำคัญของ Data Cleaning ในชีวิตจริง
