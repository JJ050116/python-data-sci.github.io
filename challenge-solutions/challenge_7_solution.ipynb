## ภารกิจ
เลือกข้อมูลหรือกราฟจาก Week 06 แล้วเขียนรายงานสรุป 3 ประโยคที่มีองค์ประกอบครบถ้วน:
1. **สิ่งที่เห็น (Observation / Answer)**
2. **หลักฐานจากข้อมูลหรือกราฟ (Evidence)**
3. **ข้อจำกัดของข้อมูล (Limitations)**

### 🔍 วิเคราะห์ Debug Checkpoint
หลีกเลี่ยงการใช้คำว่า *"พิสูจน์แล้วว่าดีที่สุดในโลก"* หรือ *"ทุกคน"* เพราะชุดข้อมูลตัวอย่างมีเพียง 47 คน การอ้างอิงเกินขอบเขตทำให้รายงานขาดความน่าเชื่อถือในทาง Data Science

```python
import pandas as pd

df = pd.read_csv("../data/footballers.csv")
top5 = df.nlargest(5, "overall")[["name", "club", "overall", "value_millions"]]

print("=== ข้อมูล Top 5 ที่นำมาใช้เป็นหลักฐาน ===")
display(top5)
```

### 📝 ตัวอย่างข้อสรุป 3 ประโยคตามเกณฑ์ประเมิน

1. **สิ่งที่เห็น (คำตอบหลัก):**
   > *"ในชุดข้อมูลนี้ Kylian Mbappé และ Erling Haaland เป็นนักเตะที่มีคะแนน Overall สูงที่สุดร่วมกันที่ 91 คะแนน"*

2. **หลักฐาน (จากตาราง/กราฟ):**
   > *"โดยคะแนน 91 นี้ สูงกว่าอันดับถัดมาอย่าง Lionel Messi (90 คะแนน) และ Vinicius Jr (89 คะแนน) อย่างชัดเจน"*

3. **ข้อจำกัดของข้อมูล (ความโปร่งใส):**
   > *"อย่างไรก็ตาม ชุดข้อมูลนี้เป็นเพียงกลุ่มตัวอย่างนักเตะชื่อดัง 47 คน จึงยังไม่สามารถสรุปภาพรวมของนักฟุตบอลอาชีพทั้งหมดในทุกลีกได้"*

```python
# โค้ดแสดงผลข้อความรายงาน
summary_report = (
    "[สรุปผลการวิเคราะห์]\n"
    f"1. ในชุดข้อมูลนี้ {top5.iloc[0]['name']} และ {top5.iloc[1]['name']} มี overall สูงสุดที่ {top5.iloc[0]['overall']} คะแนน\n"
    f"2. หลักฐาน: สูงกว่าอันดับ 3 ({top5.iloc[2]['name']}: {top5.iloc[2]['overall']} คะแนน)\n"
    f"3. ข้อจำกัด: ข้อมูลมีเพียง {len(df)} แถว จึงเป็นข้อสรุปเฉพาะกลุ่มตัวอย่างนี้"
)
print(summary_report)
```
