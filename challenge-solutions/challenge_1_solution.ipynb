## ภารกิจ
1. อ่านไฟล์ `footballers.csv` เข้าเก็บในตัวแปร `df`
2. แสดง 3 แถวแรก
3. แสดงจำนวนแถวและคอลัมน์ของตาราง (`df.shape`)

### 🔍 วิเคราะห์ Debug Checkpoint
เมื่อรันโค้ด `pd.read_csv("data/footballers.csv")` จากโฟลเดอร์ `challenge-solutions/` หรือ `challenge/` จะเกิด `FileNotFoundError`
- **สาเหตุ:** โฟลเดอร์ปัจจุบันของ Notebook อยู่ใน subfolder จึงต้องถอยกลับไป 1 ระดับด้วย `../` เพื่อเข้าถึงโฟลเดอร์ `data/`
- **การแก้ปัญหา:** เปลี่ยน path เป็น `"../data/footballers.csv"`

```python
import pandas as pd

# 1. อ่านไฟล์ด้วย relative path ที่ถูกต้อง
df = pd.read_csv("../data/footballers.csv")

# 2. แสดง 3 แถวแรก
display(df.head(3))

# 3. แสดงขนาดตาราง
print("ขนาดของตาราง (แถว, คอลัมน์):", df.shape)
print(f"-> มีข้อมูลนักเตะทั้งหมด {df.shape[0]} แถว และมี {df.shape[1]} คอลัมน์")
```

### 💡 คำอธิบายเพิ่มเติม
- `df.head(3)`: แสดงเฉพาะ 3 แถวแรกเพื่อตรวจหน้าตาและชื่อคอลัมน์ โดยไม่โหลดข้อมูลทั้งหมดออกมาให้รกหน้าจอ
- `df.shape`: คืนค่าเป็น tuple `(จำนวนแถว, จำนวนคอลัมน์)` เสมอ ทำให้รู้ขนาดข้อมูลจริงโดยไม่ต้องเดาจากสิ่งที่เห็นใน `head()`
