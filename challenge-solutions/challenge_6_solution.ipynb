## ภารกิจ
1. เลือก 5 นักเตะที่มีคะแนน `overall` สูงสุดด้วย `.nlargest(5, 'overall')`
2. สร้าง Bar Chart โดยแกน X คือชื่อนักเตะ และแกน Y คือ overall rating
3. ใส่ชื่อแกน Y (`plt.ylabel`), ชื่อกราฟ (`plt.title`) และหมุนชื่อแกน X ให้อ่านง่าย

### 🔍 วิเคราะห์ Debug Checkpoint
ถ้าส่ง `x = df["name"]` (47 คน) แต่ส่ง `y = top5["overall"]` (5 คน) จะเกิด `ValueError: x and y must have same first dimension`
- **สาเหตุ:** ขนาดของข้อมูลแกน X และ Y ไม่เท่ากัน (47 ไม่เท่ากับ 5)
- **วิธีแก้:** ดึงทั้งแกน X และ Y มาจากตารางตัวแปรเดียวกัน คือ `top5` เสมอ

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/footballers.csv")

# 1. เลือก Top 5 นักเตะ
top5 = df.nlargest(5, "overall")

# 2. สร้างกราฟแท่ง
plt.figure(figsize=(8, 5))
plt.bar(top5["name"], top5["overall"], color="#2a9d8f", edgecolor="#264653")

# 3. ตกแต่งแกนและหัวข้อกราฟ
plt.title("Top 5 Footballers by Overall Rating", fontsize=14, fontweight="bold")
plt.xlabel("Player Name", fontsize=11)
plt.ylabel("Overall Rating (0-100)", fontsize=11)
plt.ylim(80, 95)  # ขยายช่วงแกน Y เพื่อให้เห็นความต่างชัดเจนขึ้น
plt.xticks(rotation=25, ha="right")

# 4. จัดระยะห่างไม่ให้ตัวหนังสือล้นจอ
plt.tight_layout()
plt.show()
```

### 💡 คำอธิบายเพิ่มเติม
- `plt.tight_layout()` ป้องกันไม่ให้ชื่อนักเตะที่หมุนเอียงตกขอบล่างของภาพ
- การใส่หน่วยหรือช่วงคะแนนที่แกน Y ช่วยให้ผู้อ่านเข้าใจได้ทันทีว่าค่า 90 หมายถึงอะไร
